```python
import numpy as np
import matplotlib.pyplot as plt
from lafp import *

x = np.array([1.0, 2.0, -1.0])           # a vector in R^3
A = np.array([[2.0, 0.0, 1.0],
              [1.0, 3.0, 0.0]])          # a 2 x 3 matrix
print("A @ x      =", A @ x)             # a vector in R^2
print("A.T shape  =", A.T.shape)
print("A * A      =\n", A * A)           # entrywise, not matrix product
print("A @ A.T    =\n", A @ A.T)         # 2 x 2 Gram matrix of the rows
```

```python
B = np.array([[1.0, 2.0], [3.0, 4.0]])
Binv = np.linalg.inv(B)
print(B @ Binv)                           # the identity, up to rounding
print("largest deviation:", np.abs(B @ Binv - np.eye(2)).max())
print(0.1 + 0.2 == 0.3, abs(0.1 + 0.2 - 0.3))
```

A matrix acting on the plane. Left, the standard grid, unit circle and the two coordinate vectors. Right, their images under $A=\begin{bmatrix}1.2&0.8\\0.3&1.0\end{bmatrix}$. The columns of $A$ are exactly the images of the coordinate vectors.

```python
A = np.array([[1.2, 0.8], [0.3, 1.0]])
fig, axs = plt.subplots(1, 2, figsize=(W, 2.4))
for ax, M, title in [(axs[0], np.eye(2), "before"), (axs[1], A, "after $A$")]:
    plane(ax, 2.6, grid=False)
    grid_image(ax, M, lim=2, n=9, color=SKY, lw=0.5, base=False)
    circle_image(ax, M, color=RED)
    arrow(ax, M[:, 0], color=BLUE, label=r"$Ae_1$" if M is A else r"$e_1$", offset=(0.15, -0.15))
    arrow(ax, M[:, 1], color=TEAL, label=r"$Ae_2$" if M is A else r"$e_2$", offset=(-0.1, 0.2))
    ax.set_title(title)
```
